```python
columnas_categoricas = [
    'Genero',
    'Adulto_Mayor',
    'Tiene_Pareja',
    'Tiene_Dependientes',
    'Servicio_Telefonico',
    'Multiples_Lineas',
    'Servicio_Internet',
    'Tipo_Contrato',
    'Facturacion_Electronica',
    'Metodo_Pago',
    'Abandono_Cliente'
]

fig, axes = plt.subplots(4, 3, figsize=(20, 18))
axes = axes.flatten()

sns.set(style="whitegrid")

for i, columna in enumerate(columnas_categoricas):

    frecuencia = df[columna].value_counts().sort_index()

    # Cambiar etiquetas para variables binarias
    if set(df[columna].dropna().unique()) <= {0, 1}:

        frecuencia.index = [
            'No' if x == 0 else 'Sí'
            for x in frecuencia.index
        ]

    frecuencia.plot(
        kind='bar',
        ax=axes[i],
        color='skyblue'
    )

    axes[i].set_title(f'Distribución de {columna}')
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Frecuencia')

    total = frecuencia.sum()

    for j, v in enumerate(frecuencia):

        porcentaje = v / total * 100

        axes[i].text(
            j,
            v + frecuencia.max()*0.03,
            f'{v}\n({porcentaje:.1f}%)',
            ha='center',
            fontsize=9
        )

for j in range(len(columnas_categoricas), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
variables = [
    'Meses_Permanencia',
    'Cargo_Mensual',
    'Cargo_Total'
]
for variable in variables:

    resumen_grupos = df.groupby(
        'Abandono_Cliente'
    )[variable].agg(
        ['count', 'mean', 'median', 'std']
    )

    print(f'\nResumen descriptivo de {variable}')
    print(resumen_grupos.round(2))
sns.set(style="whitegrid")

for variable in variables:

    fig, ax = plt.subplots(
        1,
        2,
        figsize=(12, 5)
    )

    # Boxplot
    sns.boxplot(
        data=df,
        x='Abandono_Cliente',
        y=variable,
        palette='pastel',
        ax=ax[0]
    )

    ax[0].set_title(
        f'Boxplot de {variable} por Abandono'
    )

    ax[0].set_xlabel(
        'Abandono (0 = No, 1 = Sí)'
    )

    ax[0].set_ylabel(variable)

    # Violin Plot
    sns.violinplot(
        data=df,
        x='Abandono_Cliente',
        y=variable,
        palette='pastel',
        ax=ax[1]
    )

    ax[1].set_title(
        f'Violin Plot de {variable} por Abandono'
    )

    ax[1].set_xlabel(
        'Abandono (0 = No, 1 = Sí)'
    )

    ax[1].set_ylabel(variable)

    plt.tight_layout()
    plt.show()
```

Por su parte, Cargo_Total registró una curtosis de -0.232, valor más cercano a cero, lo que sugiere una distribución más similar a la normal respecto a las otras variables. En conjunto, estos resultados indican que las variables numéricas del conjunto de datos no presentan picos pronunciados ni una alta concentración de observaciones en torno a la media, reflejando una distribución relativamente dispersa y heterogénea de los clientes.

```python
# Limpiar espacios y convertir a número
df['Cargo_Total'] = (
    df['Cargo_Total']
    .astype(str)
    .str.strip()
)

df['Cargo_Total'] = pd.to_numeric(
    df['Cargo_Total'],
    errors='coerce'
)

# Verificar que cambió
print(df['Cargo_Total'].dtype)
```

```python
curtosis = pd.DataFrame({
    'Variable': [
        'Meses_Permanencia',
        'Cargo_Mensual',
        'Cargo_Total'
    ],
    'Curtosis': [
        df['Meses_Permanencia'].kurt(),
        df['Cargo_Mensual'].kurt(),
        df['Cargo_Total'].dropna().kurt()
    ]
})

print(curtosis)
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
La matriz de covarianza muestra que todas las relaciones entre las variables numéricas son positivas, lo que indica que cuando una variable aumenta, las demás tienden a aumentar también. La relación más fuerte se observa entre Meses_Permanencia y Cargo_Total (45.950,74), evidenciando que los clientes con mayor tiempo de permanencia acumulan un mayor valor total facturado. De igual forma, existe una covarianza positiva importante entre Cargo_Mensual y Cargo_Total (44.401,33), lo que sugiere que los clientes con cargos mensuales más elevados suelen generar mayores cargos acumulados a lo largo del tiempo. Por otro lado, la covarianza entre Meses_Permanencia y Cargo_Mensual (183,20) es considerablemente menor, indicando una asociación positiva pero más débil entre ambas variables. En general, los resultados reflejan que el Cargo_Total es la variable más relacionada con las demás, ya que depende tanto del tiempo de permanencia del cliente como de los cargos mensuales asociados a los servicios contratados.
</div>

5. Variables categóricas
</h2>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
Las tablas de frecuencia evidencian que la población analizada presenta una distribución relativamente equilibrada en algunas características demográficas, pero concentraciones importantes en variables relacionadas con los servicios contratados. En cuanto al género, la distribución es prácticamente uniforme, con un 50.48% de clientes en una categoría y un 49.52% en la otra, mientras que la mayoría de los clientes (83.79%) no corresponde a adultos mayores, representando estos únicamente el 16.21% de la muestra. Asimismo, el 51.7% de los clientes no tiene pareja y el 70.04% no tiene dependientes, lo que indica un predominio de hogares sin cargas familiares. Respecto a los servicios, se observa una alta adopción del servicio telefónico, utilizado por el 90.32% de los clientes, y entre quienes disponen de él, el 42.18% cuenta con múltiples líneas, mientras que el 48.13% posee una sola línea. En relación con el servicio de internet, la tecnología más utilizada es la fibra óptica (43.96%), seguida por DSL (34.37%), mientras que el 21.67% no dispone de conexión a internet. En cuanto a los aspectos contractuales, predominan claramente los contratos mes a mes (55.02%), seguidos por los contratos de dos años (24.07%) y un año (20.91%), lo que sugiere una importante proporción de clientes con vínculos contractuales de corto plazo. Además, el 59.22% utiliza facturación electrónica, y el método de pago más frecuente es el cheque electrónico (33.58%), seguido por cheque enviado por correo (22.89%), transferencia bancaria automática (21.92%) y tarjeta de crédito automática (21.61%). Finalmente, respecto a la variable objetivo Abandono_Cliente, se observa que el 73.46% de los clientes permanece en la compañía, mientras que el 26.54% ha abandonado el servicio. En conjunto, estos resultados muestran una base de clientes compuesta principalmente por usuarios con servicio telefónico, acceso a internet por fibra óptica, contratos mensuales y facturación electrónica, características que posteriormente resultan relevantes para comprender los factores asociados al abandono de clientes.
</div>

```python
variables = [
    'Genero',
    'Adulto_Mayor',
    'Tiene_Pareja',
    'Tiene_Dependientes',
    'Servicio_Telefonico',
    'Multiples_Lineas',
    'Servicio_Internet',
    'Tipo_Contrato',
    'Facturacion_Electronica',
    'Metodo_Pago',
    'Abandono_Cliente'
]

for variable in variables:

    tabla = pd.DataFrame(
        df[variable].value_counts()
    )

    tabla.columns = ['Frecuencia absoluta']

    # Cambiar etiquetas 0 y 1 por No y Sí en variables binarias
    if set(df[variable].dropna().unique()) <= {0, 1}:

        tabla.index = [
            'No' if valor == 0 else 'Sí'
            for valor in tabla.index
        ]

    tabla['Frecuencia relativa (%)'] = (
        tabla['Frecuencia absoluta']
        / tabla['Frecuencia absoluta'].sum()
        * 100
    ).round(2)

    tabla['Frecuencia acumulada'] = (
        tabla['Frecuencia absoluta']
        .cumsum()
    )

    tabla['Frecuencia relativa acumulada (%)'] = (
        tabla['Frecuencia relativa (%)']
        .cumsum()
    ).round(2)

    
    print(f'TABLA DE FRECUENCIAS: {variable}')


    display(tabla)
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
La matriz de correlación de las variables categóricas codificadas permitió identificar la intensidad y dirección de la relación entre las características de los clientes y la variable objetivo Abandono_Cliente. Se observa que la variable con mayor asociación es Tipo_Contrato, con una correlación negativa moderada de -0.40, indicando que a medida que los contratos son de mayor duración disminuye la probabilidad de abandono. Asimismo, Facturacion_Electronica presenta una correlación positiva de 0.19, sugiriendo una mayor tendencia al abandono entre los clientes que utilizan este tipo de facturación. Por otro lado, las variables Tiene_Dependientes (-0.16) y Tiene_Pareja (-0.15) muestran relaciones negativas con el abandono, lo que indica que los clientes con estas características tienden a permanecer más tiempo en la compañía, mientras que Adulto_Mayor presenta una correlación positiva de 0.15, evidenciando una mayor propensión al churn en este grupo. En contraste, variables como Genero (-0.01), Servicio_Telefonico (0.01), Multiples_Lineas (0.04) y Servicio_Internet (-0.05) presentan correlaciones muy cercanas a cero, por lo que su influencia sobre el abandono parece ser limitada. Adicionalmente, la matriz permitió evaluar la posible presencia de multicolinealidad entre las variables independientes, observándose que la mayoría de las correlaciones se encuentran por debajo de |0.50|. Las asociaciones más destacadas corresponden a Tiene_Pareja y Tiene_Dependientes (0.45), así como Servicio_Telefonico y Servicio_Internet (0.39), las cuales indican una relación moderada pero insuficiente para generar problemas de redundancia en futuros modelos predictivos. Dado que ninguna correlación supera niveles cercanos a 0.70 o 0.80, no se evidencian problemas significativos de multicolinealidad. En general, los resultados coinciden con el análisis bivariado realizado previamente y confirman que las variables relacionadas con las condiciones contractuales y los servicios contratados tienen una mayor influencia sobre el abandono de clientes que las variables demográficas.
</div>

```python
categoricas = [
    'Genero',
    'Adulto_Mayor',
    'Tiene_Pareja',
    'Tiene_Dependientes',
    'Servicio_Telefonico',
    'Multiples_Lineas',
    'Servicio_Internet',
    'Tipo_Contrato',
    'Facturacion_Electronica',
    'Metodo_Pago',
    'Abandono_Cliente'
]

# Crear copia
df_corr = df[categoricas].copy()

# Convertir variables categóricas a códigos numéricos
for col in df_corr.columns:
    df_corr[col] = df_corr[col].astype('category').cat.codes

# Matriz de correlación
matriz_corr = df_corr.corr()

# Graficar
plt.figure(figsize=(12,8))

sns.heatmap(
    matriz_corr,
    annot=True,
    fmt='.2f',
    cmap='Pastel1',
    linewidths=0.5
)

plt.title('Matriz de Correlación de Variables Categóricas')
plt.tight_layout()
plt.show()
```

<h2 style="color:#00A86B; font-family:Arial; border-bottom:3px solid #00A86B; padding-bottom:8px;">
6. Análisis bivariado de la variable objetivo churn
</h2>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
El análisis bivariado permitió identificar que las variables con mayor relación con el abandono de clientes son el tipo de contrato, el servicio de internet, la facturación electrónica y el método de pago. Los clientes con contratos mes a mes presentan la mayor tasa de abandono (42.7%), mientras que aquellos con contratos de uno y dos años muestran porcentajes significativamente menores. Asimismo, los usuarios de fibra óptica registran una mayor proporción de abandono (41.9%) en comparación con los clientes de DSL o sin servicio de internet. También se observa que quienes utilizan facturación electrónica abandonan con mayor frecuencia (33.6%) que quienes no la utilizan (16.3%). Respecto al método de pago, el cheque electrónico presenta la tasa de abandono más elevada (45.3%), mientras que los pagos automáticos muestran los menores porcentajes. Por otro lado, variables como género, servicio telefónico y múltiples líneas presentan diferencias poco significativas, por lo que parecen tener una influencia limitada sobre el churn. En general, los resultados sugieren que las condiciones contractuales y de servicio son los principales factores asociados al abandono de clientes.
</div>

```python
sns.set(style="whitegrid")

variables = [
    'Genero',
    'Adulto_Mayor',
    'Tiene_Pareja',
    'Tiene_Dependientes',
    'Servicio_Telefonico',
    'Multiples_Lineas',
    'Servicio_Internet',
    'Tipo_Contrato',
    'Facturacion_Electronica',
    'Metodo_Pago'
]

fig, axes = plt.subplots(5, 2, figsize=(20, 25))
axes = axes.flatten()

for i, variable in enumerate(variables):

    tabla = pd.crosstab(
        df[variable],
        df['Abandono_Cliente']
    )

    ax = tabla.plot(
        kind='bar',
        ax=axes[i],
        color=['#A8DADC', '#F4A6A6'],  # Azul pastel y salmón pastel
        width=0.8
    )

    ax.set_title(
        f'{variable} vs Abandono_Cliente',
        fontsize=12,
        pad=10
    )

    ax.set_xlabel(variable)
    ax.set_ylabel('Frecuencia')
    totales_categoria = tabla.sum(axis=1)

    for container in ax.containers:

        for j, barra in enumerate(container):

            altura = barra.get_height()

            if altura > 0:

                porcentaje = (
                    altura /
                    totales_categoria.iloc[j]
                ) * 100

                ax.text(
                    barra.get_x() + barra.get_width()/2,
                    altura,
                    f'{int(altura)}\n({porcentaje:.1f}%)',
                    ha='center',
                    va='bottom',
                    fontsize=8
                )

    ax.legend(
        title='Abandono Cliente',
        labels=['No', 'Sí']
    )

    etiquetas = []

    for tick in ax.get_xticklabels():

        texto = tick.get_text()

        if texto == '0':
            etiquetas.append('No')
        elif texto == '1':
            etiquetas.append('Sí')
        else:
            etiquetas.append(texto)

    ax.set_xticklabels(
        etiquetas,
        rotation=30,
        ha='right'
    )

plt.tight_layout()
plt.show()
```

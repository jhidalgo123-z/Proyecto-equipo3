</div>

| Pregunta | Respuesta del equipo |
|---|---|
| ¿Qué **decisión** de negocio apoya el modelo? | **A qué clientes ofrecer una campaña de retención** (descuento, mejora de plan o llamada del equipo de fidelización) antes de que cancelen |
| ¿**Quién** toma esa decisión? | La gerencia de retención y fidelización (área de *Customer Success*) de la empresa de telecomunicaciones |
| ¿Qué representa **una fila**? | Un cliente, con sus datos demográficos, servicios contratados, contrato y facturación |
| ¿Qué **tipo de análisis** es el principal? | **Predictivo** (clasificación binaria), apoyado en el EDA diagnóstico |

| **KPI 1, de negocio (SMART)** | Reducir la tasa de abandono del **26,5 % al 22 %** en los **6 meses** siguientes a la puesta en marcha, con campañas de retención dirigidas solo a los clientes que el modelo marque en riesgo |
| **KPI 2, del modelo = criterio de éxito (SMART)** | En el conjunto de **test** (20 % de los clientes, nunca visto en el entrenamiento), detectar **al menos el 75 %** de los clientes que abandonan (**recall de la clase 1 ≥ 0,75**) con **precisión ≥ 0,50** y **ROC-AUC ≥ 0,80**, y superar al *baseline* |

<h3 style="color:#00A86B; font-family:Arial;">¿Por qué el recall es la métrica principal y no la accuracy?</h3>

Cada predicción del modelo cae en una de cuatro casillas. En lenguaje de negocio:

| | El modelo dice: **se queda** | El modelo dice: **abandona** |
|---|---|---|
| **Realmente se quedó** | TN: no hacemos nada y estaba bien | **FP (falsa alarma):** le damos un descuento a alguien que igual se iba a quedar |
| **Realmente abandonó** | **FN (se escapó):** perdemos al cliente y todo su ingreso futuro sin intentar retenerlo | TP: lo detectamos a tiempo y podemos intentar retenerlo |

- **Un FN cuesta mucho más que un FP.** Un cliente perdido se lleva todos sus pagos futuros, y conseguir uno nuevo cuesta más que retenerlo. Una falsa alarma solo cuesta un descuento o una llamada. Por eso la métrica principal es el **recall de la clase "abandona"**: de los clientes que se van, ¿cuántos detectamos?
- **La precisión es una restricción de costo.** Si marcamos a todos como "en riesgo", el recall llega a 100 %, pero la campaña se vuelve carísima. Exigir **precisión ≥ 0,50** significa que al menos 1 de cada 2 clientes contactados sí se iba a ir.
- **ROC-AUC y F1 son secundarias.** El ROC-AUC mide qué tan bien *ordena* el modelo a los clientes por riesgo, sin depender del umbral, y sirve para comparar modelos. El F1 resume el balance entre recall y precisión.
- **La accuracy NO se usa.** Con 73,5 % de clientes que se quedan, un "modelo" que dice siempre "se queda" tiene **73,5 % de accuracy** y no detecta a **ningún** cliente que abandona (recall = 0). La accuracy premia exactamente lo que no nos sirve.

**¿Por qué 0,75 y 0,50?** Los umbrales se fijaron antes de modelar y después se revisó que fueran alcanzables: con una regresión logística sin ajustar, la validación cruzada da recall ≈ 0,79 y precisión ≈ 0,52. La meta es exigente pero realista. Pedir recall ≥ 0,85 obligaría a bajar la precisión a ≈ 0,47 (ver la tabla de umbrales de la sección 9.6).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.inspection import permutation_importance
from sklearn.metrics import classification_report
from sklearn.model_selection import (GridSearchCV, StratifiedKFold, cross_val_predict,
                                     cross_validate, train_test_split)

from proyecto_equipo_3 import metricas as mt
from proyecto_equipo_3 import modelo as md

CRITERIO = md.CRITERIO_EXITO
print("Criterio de éxito (KPI 2):", CRITERIO)
```

<h3 style="color:#00A86B; font-family:Arial;">9.1 Selección de variables</h3>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
Quitamos <code>ID_Cliente</code> porque es un identificador. Antes de decidir qué hacer con <code>Cargo_Total</code> revisamos la multicolinealidad: si el cargo total es básicamente <i>meses × cargo mensual</i>, repite información que el modelo ya tiene.
</div>

```python
producto = df["Meses_Permanencia"] * df["Cargo_Mensual"]
print(f"corr(Cargo_Total, Meses x Cargo_Mensual) = {df['Cargo_Total'].corr(producto):.4f}")
print(f"corr(Cargo_Total, Meses_Permanencia)     = {df['Cargo_Total'].corr(df['Meses_Permanencia']):.4f}")
print(f"corr(Cargo_Total, Cargo_Mensual)         = {df['Cargo_Total'].corr(df['Cargo_Mensual']):.4f}")

# ¿Aporta algo Cargo_Total? Misma regresión logística con y sin la variable (validación cruzada en todo el dataset)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEMILLA)
X_ct = df[md.FEATURES + ["Cargo_Total"]]
comparacion = {}
for nombre, numericas in [("sin Cargo_Total", md.VARIABLES_NUMERICAS),
                          ("con Cargo_Total", md.VARIABLES_NUMERICAS + ["Cargo_Total"])]:
    pipe = md.construir_pipeline(md.modelos_candidatos()["Regresión logística"], numericas)
    res = cross_validate(pipe, X_ct, df[md.OBJETIVO], cv=cv, scoring=["recall", "roc_auc"])
    comparacion[nombre] = {"recall_cv": res["test_recall"].mean().round(3), "roc_auc_cv": res["test_roc_auc"].mean().round(3)}
pd.DataFrame(comparacion).T
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
<b>Decisión: Cargo_Total NO entra al modelo.</b> Su correlación con <i>Meses_Permanencia × Cargo_Mensual</i> es de 0,9996: es prácticamente la misma información. Incluirla casi no cambia el resultado (recall 0,794 frente a 0,802 y ROC-AUC 0,843 frente a 0,845, diferencias menores que la variación entre folds, que es de ±0,03 en recall) y, en la regresión logística, dos variables casi idénticas vuelven inestables los coeficientes y más difícil explicar el modelo. Además se evita imputar los 11 blancos.
</div>

**Revisión de *data leakage*.** Una variable filtra la respuesta si solo se conoce *después* de que el cliente se va. Ninguna de las 18 variables que usamos es de ese tipo: son datos del perfil, del contrato y de la facturación que la empresa tiene **mientras el cliente sigue activo**. Además, todo el preprocesamiento (mediana, escalado, *one-hot*) se ajusta **solo con train** dentro del pipeline. Una advertencia: en este dataset, `Meses_Permanencia` de quien ya se fue es su permanencia final. En producción se usa la permanencia *actual* del cliente activo, lo cual es coherente, pero conviene vigilarlo (ver límites).

| Tipo | Variables | Tratamiento en el pipeline |
|---|---|---|
| Numéricas (2) | `Meses_Permanencia`, `Cargo_Mensual` | Imputar con la mediana + `StandardScaler` |
| Binarias 0/1 (6) | `Genero`, `Adulto_Mayor`, `Tiene_Pareja`, `Tiene_Dependientes`, `Servicio_Telefonico`, `Facturacion_Electronica` | Imputar con la moda + `OneHotEncoder` |
| Categóricas (10) | `Multiples_Lineas`, `Servicio_Internet`, los 6 servicios adicionales, `Tipo_Contrato`, `Metodo_Pago` | Imputar con la moda + `OneHotEncoder(handle_unknown="ignore")` |

<h3 style="color:#00A86B; font-family:Arial;">9.2 Separar entrenamiento y prueba</h3>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
80 % para entrenar y 20 % para la evaluación final. Usamos <code>stratify=y</code> para que ambos conjuntos tengan la misma proporción de clientes que abandonan: con clases desbalanceadas, un corte al azar podría dejar el test con más o menos abandonos de los reales. El test se usa <b>una sola vez</b>, al final. Para comparar modelos y elegir hiperparámetros usamos <b>validación cruzada de 5 folds</b> dentro de train, que actúa como nuestro conjunto de validación.
</div>

```python
X, y = md.separar_x_y(df)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=SEMILLA
)
pd.DataFrame({
    "clientes": [len(y_train), len(y_test)],
    "abandonan": [y_train.sum(), y_test.sum()],
    "% abandono": [round(y_train.mean() * 100, 1), round(y_test.mean() * 100, 1)],
}, index=["train", "test"])
```

<h3 style="color:#00A86B; font-family:Arial;">9.3 Baseline y modelos de complejidad creciente (validación cruzada)</h3>

| Modelo | En palabras simples |
|---|---|
| **Baseline** (`DummyClassifier`, clase más frecuente) | Dice siempre "se queda". Es el piso: cualquier modelo útil debe superarlo |
| **Regresión logística** | Suma ponderada de las variables convertida en probabilidad. Es simple, rápida e interpretable (cada variable tiene un coeficiente) |
| **Árbol de decisión** (profundidad 5) | Preguntas sí/no encadenadas ("¿contrato mes a mes?", "¿menos de 6 meses?") |
| **Random Forest** | Cientos de árboles que votan. Es más robusto, pero tiende a memorizar |
| **Gradient Boosting** | Árboles en secuencia, cada uno corrige los errores del anterior |

Todos usan `class_weight="balanced"` (excepto el baseline), que le da más peso a la clase minoritaria para que el modelo no "se rinda" y prediga siempre "se queda". Comparamos con **recall** como métrica principal y además precision, F1, **F2** (que da el doble de peso al recall) y ROC-AUC.

```python
resultados_cv = {}
for nombre, estimador in md.modelos_candidatos().items():
    resultados_cv[nombre] = cross_validate(
        md.construir_pipeline(estimador), X_train, y_train, cv=cv, scoring=mt.SCORING_CV, n_jobs=-1
    )
tabla_cv = mt.resumen_cv(resultados_cv)
tabla_cv[["recall_cv", "recall_std", "precision_cv", "f1_cv", "f2_cv", "roc_auc_cv", "roc_auc_std"]]
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
<b>Lectura:</b> el <i>baseline</i> tiene recall 0: no detecta a ningún cliente que abandona. Todos los modelos le ganan con claridad. La <b>regresión logística</b> queda primera en recall (≈ 0,79), F2 y ROC-AUC (≈ 0,845), seguida muy de cerca por el árbol. Los <i>ensembles</i> (Random Forest y Gradient Boosting), aunque son más complejos, detectan <b>menos</b> clientes que abandonan. Más complejidad no significó mejor resultado. La desviación estándar entre folds (≈ 0,03 en recall) muestra que los resultados son estables.
</div>

<h3 style="color:#00A86B; font-family:Arial;">9.4 ¿Overfitting o underfitting? Train frente a test</h3>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
Entrenamos cada modelo con todo train y medimos en train <b>y</b> en test (umbral 0,5). Si el modelo es mucho mejor en train que en test, memorizó (<i>overfitting</i>). Si es malo en ambos, es demasiado simple (<i>underfitting</i>).
</div>

```python
modelos_entrenados = {
    nombre: md.construir_pipeline(estimador).fit(X_train, y_train)
    for nombre, estimador in md.modelos_candidatos().items()
}
tabla_tt = mt.tabla_train_test(modelos_entrenados, X_train, y_train, X_test, y_test)
tabla_tt[["recall_train", "recall_test", "precision_train", "precision_test",
          "roc_auc_train", "roc_auc_test", "accuracy_test", "brecha_recall", "brecha_auc"]]
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
<ul><li><b>Random Forest: overfitting claro.</b> Recall de 1,00 en train y de 0,61 en test (brecha de 0,39): memorizó a los clientes de entrenamiento. Gradient Boosting también se sobreajusta (0,94 frente a 0,75).</li><li><b>Regresión logística y árbol: buen ajuste.</b> Brechas de 0,02 a 0,03 en recall: lo que aprendieron en train se mantiene con clientes nuevos.</li><li><b>Baseline: underfitting extremo.</b> No aprende nada. Su accuracy de 0,735 parece buena, pero su recall es 0. Es la paradoja de la accuracy en vivo.</li></ul>Con la regla de oro (el mejor en test sin descuidar train), pasan a la búsqueda de hiperparámetros los dos mejores en validación cruzada y con poca brecha: <b>regresión logística</b> y <b>árbol de decisión</b>.
</div>

<h3 style="color:#00A86B; font-family:Arial;">9.5 Ajuste de hiperparámetros con GridSearchCV</h3>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
<code>GridSearchCV</code> prueba todas las combinaciones de la grilla con validación cruzada de 5 folds (solo sobre train). Como criterio para elegir (<code>refit</code>) usamos <b>F2</b> y no el recall solo: optimizar únicamente el recall premia al modelo que marca a todos como "en riesgo" (recall 1, precisión pésima). F2 combina ambos dándole el doble de peso al recall, que es justo el costo relativo del negocio.
</div>

```python
busquedas = {}
for nombre in ["Regresión logística", "Árbol de decisión"]:
    busquedas[nombre] = GridSearchCV(
        md.construir_pipeline(md.modelos_candidatos()[nombre]),
        md.GRILLAS[nombre], cv=cv, scoring=mt.SCORING_CV, refit="f2", n_jobs=-1,
    ).fit(X_train, y_train)

filas = []
for nombre, busqueda in busquedas.items():
    i, r = busqueda.best_index_, busqueda.cv_results_
    filas.append({
        "modelo": nombre,
        "mejores_hiperparametros": {k.replace("modelo__", ""): v for k, v in busqueda.best_params_.items()},
        **{f"{m}_cv": round(r[f"mean_test_{m}"][i], 3) for m in ["recall", "precision", "f2", "roc_auc"]},
    })
pd.DataFrame(filas).set_index("modelo")
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
<b>Elegimos la regresión logística</b> (C = 0,01, <i>solver</i> liblinear). En validación cruzada queda prácticamente empatada con el árbol (F2 0,720 frente a 0,718, una diferencia menor que la variación entre folds), pero tiene mayor ROC-AUC (0,845 frente a 0,833), menor brecha entre train y test, y sus coeficientes se pueden explicar al comité directivo. Una C pequeña significa <b>más regularización</b>: el modelo es más "prudente" y generaliza mejor. Ante un empate técnico, preferimos el modelo más simple e interpretable.
</div>

```python
modelo_final = busquedas["Regresión logística"].best_estimator_
hiperparametros = {k.replace("modelo__", ""): v for k, v in busquedas["Regresión logística"].best_params_.items()}
assert hiperparametros == md.HIPERPARAMETROS_FINALES, "Actualizar HIPERPARAMETROS_FINALES en modelo.py"
print("Modelo final:", md.NOMBRE_MODELO_FINAL, hiperparametros)
```

<h3 style="color:#00A86B; font-family:Arial;">9.6 Elección del umbral de decisión</h3>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
El modelo entrega una <b>probabilidad</b> de abandono, y el umbral decide a partir de qué probabilidad contactamos al cliente. Es una <b>decisión de negocio</b>, no técnica. Para no "gastar" el test, elegimos el umbral con predicciones de <b>validación cruzada sobre train</b> (cada cliente es predicho por un modelo que no lo vio). La regla sale del KPI: <b>el umbral con mayor recall entre los que mantienen la precisión ≥ 0,50</b>.
</div>

```python
prob_validacion = cross_val_predict(modelo_final, X_train, y_train, cv=cv, method="predict_proba")[:, 1]
tabla_umbral_val = mt.tabla_umbrales(y_train, prob_validacion)
UMBRAL = mt.elegir_umbral(tabla_umbral_val, CRITERIO["precision"])
print(f"Umbral elegido con validación cruzada: {UMBRAL}")
tabla_umbral_val.loc[[0.3, 0.4, 0.45, 0.5, 0.55, 0.6, 0.7]]
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
Con 0,45 la precisión baja de 0,50 (demasiadas falsas alarmas). Con 0,55 o más, la precisión sube, pero se escapan más clientes. El umbral <b>0,5</b> es el punto que maximiza la detección respetando la restricción de costo. Ahora sí, evaluamos <b>una sola vez</b> en test.
</div>

<h3 style="color:#00A86B; font-family:Arial;">9.7 Evaluación final en test: matriz de confusión</h3>

```python
prob_test = modelo_final.predict_proba(X_test)[:, 1]
pred_test = (prob_test >= UMBRAL).astype(int)
tn, fp, fn, tp = mt.matriz_confusion(y_test, pred_test, f"Regresión logística en test (umbral {UMBRAL})")

print(f"De {tp + fn} clientes que realmente se fueron, el modelo detectó {tp} ({tp / (tp + fn):.1%}) "
      f"y se le escaparon {fn}.")
print(f"El modelo marcó a {tp + fp} clientes en riesgo: {tp} sí se iban a ir y {fp} eran falsas alarmas "
      f"(precisión {tp / (tp + fp):.1%}).")
print(f"De {tn + fp} clientes que se quedaron, {tn} fueron bien identificados como 'se queda'.")
print(f"El baseline habría detectado 0 de {tp + fn}.")
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
<b>En lenguaje de negocio:</b> de los <b>374</b> clientes del test que se fueron, el modelo detectó <b>295 (78,9 %)</b> y se le escaparon <b>79</b> (FN). Para encontrarlos, la campaña contactaría a <b>578</b> clientes: 295 sí se iban a ir (TP) y <b>283</b> son falsas alarmas (FP) que recibirían un beneficio sin necesitarlo. Sin modelo (<i>baseline</i>) no se detecta a ninguno, y contactar a todos costaría 1.409 contactos para encontrar a los mismos 374.

<h3 style="color:#00A86B; font-family:Arial;">9.11 ¿Qué variables pesan más?</h3>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
Dos miradas complementarias. Los <b>coeficientes</b> de la regresión logística (las variables están escaladas, así que son comparables): positivo aumenta el riesgo y negativo lo disminuye. La <b>importancia por permutación</b> en test: cuánto empeora el AUC si "revolvemos" una variable.
</div>

```python
nombres = modelo_final[:-1].get_feature_names_out()
coeficientes = pd.Series(modelo_final[-1].coef_[0], index=nombres)
coeficientes.index = coeficientes.index.str.replace("numericas__", "").str.replace("categoricas__", "")
extremos = pd.concat([coeficientes.nsmallest(8), coeficientes.nlargest(8)]).sort_values()

importancia = permutation_importance(modelo_final, X_test, y_test, scoring="roc_auc",
                                     n_repeats=10, random_state=SEMILLA, n_jobs=-1)
importancia = pd.Series(importancia.importances_mean, index=X_test.columns).sort_values()

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
extremos.plot.barh(ax=ax[0], color=np.where(extremos > 0, "#C2262B", "#00A86B"))
ax[0].set_title("Coeficientes: rojo aumenta el riesgo, verde lo reduce")
importancia.tail(10).plot.barh(ax=ax[1], color="#00A86B")
ax[1].set_title("Importancia por permutación (caída del AUC en test)")
plt.tight_layout()
plt.show()
importancia.sort_values(ascending=False).head(6).round(4)
```

<div style="max-width:900px; text-align:justify; line-height:1.7;">
El modelo confirma lo que vimos en el EDA. Lo que más pesa es la <b>permanencia</b>: cada mes adicional reduce el riesgo (H2). Después viene el <b>tipo de contrato</b>: el mes a mes es el coeficiente que más aumenta el riesgo y el de 2 años el que más lo reduce (H1). Les siguen la <b>fibra óptica</b>, el <b>cargo mensual</b> y el <b>pago con cheque electrónico</b> (H3). Tener <b>seguridad en línea y soporte técnico</b> reduce el riesgo, lo que da una idea para la campaña: ofrecer esos servicios. <b>Género</b>, pareja y <i>streaming</i> no aportan casi nada (importancia ≈ 0), igual que en el EDA.
</div>

<h2 style="color:#00A86B; font-family:Arial; border-bottom:3px solid #00A86B; padding-bottom:8px;">
11. Conclusiones para el negocio
</h2>

<div style="max-width:900px; text-align:justify; line-height:1.7;">
Traducimos el resultado del test a la escala de toda la base y calculamos qué tendría que lograr la campaña para cumplir el KPI 1 (bajar el abandono del 26,5 % al 22 %).
</div>

```python
clientes = len(df)
abandonan = int(df[md.OBJETIVO].sum())
meta_abandono = 0.22
recall_test, precision_test = metricas_test["recall"], metricas_test["precision"]

clientes_a_retener = abandonan - meta_abandono * clientes
detectados = recall_test * abandonan
contactados = detectados / precision_test
exito_necesario = clientes_a_retener / detectados

print(f"Base: {clientes:,} clientes · {abandonan:,} abandonan ({abandonan / clientes:.1%})")
print(f"Para llegar al 22 % hay que retener a {clientes_a_retener:,.0f} clientes")
print(f"Con recall {recall_test}, el modelo detectaría ≈ {detectados:,.0f} de los {abandonan:,} que se van")
print(f"La campaña contactaría ≈ {contactados:,.0f} clientes ({contactados / clientes:.0%} de la base)")
print(f"→ Se necesita convencer a ≈ {exito_necesario:.0%} de los clientes en riesgo detectados")
```

**Respuesta al problema.** Sí es posible anticipar el abandono con los datos que la empresa ya tiene. El modelo permite decidir **a quién** ofrecer la campaña de retención, en lugar de contactar a todos o a nadie.

**Resultado principal en unidades de negocio.**
- **De cada 100 clientes que se van, el modelo detecta 79 a tiempo; sin modelo no se detecta ninguno.**
- De cada 100 clientes que el modelo marca en riesgo, **51 sí se iban a ir**. Los otros 49 reciben un beneficio que no necesitaban (es el costo de la campaña).
- Para cumplir el KPI 1 (bajar del 26,5 % al 22 %), la campaña debe retener a **cerca de 1 de cada 5 clientes detectados** (≈ 22 %). Este dato se debe validar con un piloto, porque **no tenemos información sobre la efectividad real de las campañas**.
- Los factores de riesgo coinciden en el EDA y en el modelo: **contrato mes a mes, pocos meses de permanencia, fibra óptica, cargos altos y cheque electrónico**. Los servicios de **seguridad en línea y soporte técnico** se asocian con menor riesgo.

**Límites (cuándo NO confiar en el modelo).**
1. **Casi no detecta el abandono en contratos de 1 y 2 años ni en clientes sin internet.** Para esos grupos, la empresa necesita otra estrategia (por ejemplo, alertas antes del vencimiento del contrato).
2. **La precisión está justo en el límite (0,51):** cerca de la mitad de los contactos son falsas alarmas. Si la campaña es costosa, conviene subir el umbral.
3. **Es una foto de un solo momento**, sin fechas: no sabemos *cuándo* se irá el cliente ni si los patrones cambian con el tiempo (*data drift*). Hay que reentrenar periódicamente.
4. **Asociación no es causalidad:** que la fibra se asocie con abandono no prueba que la fibra lo cause (puede ser el precio o la competencia).
5. Dataset público de una empresa ficticia (IBM/Kaggle). Antes de usarlo se debe validar con datos propios de la operación.

**Siguientes pasos.**
1. **Piloto A/B de 3 meses:** a la mitad de los clientes en riesgo se les ofrece la campaña y a la otra mitad no. Así se mide el efecto real sobre el abandono.
2. Estimar el **costo real de un FN** (ingreso perdido) **y de un FP** (costo del beneficio) para fijar el umbral por utilidad esperada.
3. Agregar variables de **comportamiento** (llamadas a soporte, quejas, uso, fecha de fin de contrato) para mejorar los contratos largos.
4. **Monitorear** mensualmente el recall y la tasa de abandono, y reentrenar con `scripts/train.py --buscar --version 1.1.0`.
5. Integrar la **API** (`/predict-batch`) en el CRM para generar la lista mensual de clientes a contactar.
